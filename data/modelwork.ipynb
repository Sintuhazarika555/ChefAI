# PREPROCESSING


Inputs:ingredients, equipment, and maximum stated time.

Targets: food name (title) and method (directions).

- Use the existing NER field as the ingredient-entity list.
- Extract equipment and stated durations from directions with a baseline rule-based parser.
- NER contains ingredients, not taste labels. Taste needs a separate labeled source or an explicit flavor vocabulary/classifier; it is not inferred in this preprocessing step.


```python
df.head(1)
```

```python
import ast

"Parses stringified python lists into clean, space-separated plain text."
def clean_sequence(col_val) -> str:
    if pd.isna(col_val):
        return ""
    try:
        items = ast.literal_eval(col_val)
        if isinstance(items, list):
            return " ".join([str(i).strip() for i in items])
        return str(col_val).strip()
    except Exception:
        return str(col_val).strip()

df["clean_ingredients"] = df["ingredients"].apply(clean_sequence)
df["clean_directions"] = df["directions"].apply(clean_sequence)
```

```python
df.head()
```

```python
"Extracts total cooking time in minutes from directions text."
import re 

def parse_cooking_time(text: str) -> int:
    
    if not isinstance(text, str) or not text:
        return 30
    total = 0
    # Capture hours (e.g. '1 hr', '2 hours', '1.5 hrs')
    for hr in re.findall(r"(\d+(?:\.\d+)?)\s*(?:hr|hour)s?", text, re.I):
        total += int(float(hr) * 60)
    # Capture minutes or ranges (e.g. '25 min', '30-35 mins')
    for lower, upper in re.findall(r"(?:(\d+)\s*-\s*)?(\d+)\s*(?:min|minute)s?", text, re.I):
        total += int(upper if upper else lower)
    # Reasonable fallback clamp
    return total if 5 <= total <= 480 else 30


df["cook_time_mins"] = df["clean_directions"].apply(parse_cooking_time)
```

```python
df.head(1)
```

```python
"for tool and appliance mentions."

EQUIPMENT= {
    "oven": re.compile(r"\b(oven|bake|baking|baked|roast|roasting|preheat|broil)\b", re.I),
    "blender": re.compile(r"\b(blender|blend|puree|purée|food processor)\b", re.I),
    "stovetop": re.compile(r"\b(skillet|pan|saucepan|pot|boil|simmer|saute|sauté|fry|frying|sear)\b", re.I),
    "microwave": re.compile(r"\b(microwave|microwaving|microwaved)\b", re.I),
    "slow_cooker": re.compile(r"\b(slow cooker|crockpot|crock-pot|instant pot)\b", re.I)
}

def detect_equipment(text: str) -> list:
    if not isinstance(text, str):
        return ["stovetop"]
    found = [eq for eq, pattern in EQUIPMENT.items() if pattern.search(text)]
    return found if found else ["stovetop"]

df["equipment"] = df["clean_directions"].apply(detect_equipment)
```

```python
df.head(1)
```

```python
"for taste extraction"

TASTE= {
    "Sweet": ["sugar", "vanilla", "honey", "chocolate", "maple", "caramel", "cinnamon", "brown sugar"],
    "Spicy": ["jalapeno", "chili", "cayenne", "sriracha", "habanero", "red pepper flakes", "curry"],
    "Tangy": ["lemon", "lime", "vinegar", "mustard", "pickle", "yogurt"],
    "Umami": ["soy sauce", "mushroom", "parmesan", "worcestershire", "miso", "beef broth"],
    "Savory": ["garlic", "onion", "butter", "olive oil", "rosemary", "thyme", "black pepper"]
}

def assign_seed_taste(text: str) -> str:
    text = text.lower()
    scores = {taste: sum(seed in text for seed in seeds) for taste, seeds in TASTE.items()}
    best_taste = max(scores, key=scores.get)
    return best_taste if scores[best_taste] > 0 else "Savory"

df["taste_profile"] = df["clean_ingredients"].apply(assign_seed_taste)
```

```python
# here we are doing two things  recommendation system for food recommend .
# and classification for classifying flavor of different food . 
```

```python
#working with taste
import os
import pickle
from pathlib import Path
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(df["taste_profile"])

with open("artifacts/label_encoder.pkl","wb") as f:
    pickle.dump(label_encoder,f)

print("Classes mapped:", dict(zip(label_encoder.classes_, range(len(label_encoder.classes_)))))
print("Saved artifacts/label_encoder.pkl successfully!")
```

```python
#working with ingredient encoding
import joblib
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf_vectorizer = TfidfVectorizer(
    ngram_range=(1,2),
    max_features=5000,
    stop_words="english",
    sublinear_tf=True
)

X_tfidf = tfidf_vectorizer.fit_transform(df["clean_ingredients"])

joblib.dump(tfidf_vectorizer,"artifacts/tfidf_recommender.pkl")
print(f"TF-IDF Matrix shape: {X_tfidf.shape}")
print("Saved artifacts/tfidf_recommender.pkl successfully!")
```

# MODEL PART

```python
# we use MLP Classifier (Multi-Layer Perceptron Classifier)
```

```python
from sklearn.neural_network import MLPClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
```

```python
x_train,x_test,y_train,y_test= train_test_split(
    X_tfidf, y_encoded , test_size=0.15 , random_state=42, stratify = y_encoded
)
```

```python
mlp = MLPClassifier(
    hidden_layer_sizes =(64,32),
    activation="relu",
    solver="adam",
    max_iter=10,
    random_state=42,
    verbose=True
)
```

```python
mlp.fit(x_train,y_train)
```

```python
y_pred = mlp.predict(x_test)
print(classification_report(y_test, y_pred, target_names=label_encoder.classes_))
```

```python
#testing 
test_input = ["dark chocolate cocoa powder sugar vanilla butter"]
test_vec = tfidf_vectorizer.transform(test_input)

predicted_idx = mlp.predict(test_vec)[0]
predicted_taste = label_encoder.inverse_transform([predicted_idx])[0]

print("Test Ingredients:", test_input[0])
print("Predicted Taste:", predicted_taste)
```

```python
#making artifact

joblib.dump(mlp, "artifacts/taste_dl_model.pkl")
print("Saved artifacts/taste_dl_model.pkl successfully!")
```
